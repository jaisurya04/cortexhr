# Stage 7 of 12: Train-Test Split
Stratified split first; then every learned number comes from TRAINING rows only.

```python
Xtr, ytr, Xte, yte, folds = S['Xtr'], S['ytr'], S['Xte'], S['yte'], S['folds']
print('X:', S['X'].shape, '| target in X?', C.TARGET in S['X'].columns, '| train', Xtr.shape, 'test', Xte.shape)
pipe = F.build_pipeline(True).fit(Xtr, ytr)               # fit sees training rows only
d = pipe.named_steps['derive']
print('department-pay medians learned from', len(Xtr), 'training rows:', len(d.grp_median_), 'dept+level groups | global median pay =', d.global_median_)
print('the', len(Xte), 'test rows influenced none of these numbers')
```

**Why fit on training data only (3 lines).** 1) Medians, means, categories and department pay medians describe the data; computing them on test rows leaks test information into training. 2) A new employee scored tomorrow is not in the data the pipeline learned from, so the honest test is data it never saw. 3) Putting the pipeline inside `Pipeline.fit()` (and inside every CV fold) guarantees it.

# Stage 10 of 12: Hyperparameter Tuning
Which aggregation window works best? (3 months was the brief's example.) Evaluated on rows that have monthly history, CV only.

```python
from sklearn.model_selection import StratifiedKFold
rows = []
for w in [1, 3, 6]:
    jw, _ = D.join_monthly_features(prepared, att, pay, window=w); Xw, yw = F.make_xy(jw, True)
    mk_idx = jw.att_months_covered.notna().to_numpy(); Xm, ym = Xw[mk_idx], yw[mk_idx]
    ff = list(StratifiedKFold(5, shuffle=True, random_state=C.SEED).split(Xm, ym))
    s = cross_val_predict(mk(LRB(), True), Xm, ym, cv=ff, method='predict_proba')[:, 1]
    rows.append({'window_months': w, 'rows': len(Xm), 'leavers': int(ym.sum()), 'cv_roc_auc': round(roc_auc_score(ym, s), 3)})
pd.DataFrame(rows)
```

Only ~380 rows (about 50 leavers) are available, so differences between windows are noise-level; the brief's 3-month window is kept.

## A2. `02_feature_pipeline.ipynb` (Task 2)

### Stage 7 — Why fit on training only? (3 lines)

1. Medians, means, categories and department pay medians describe the data; if I compute them on test rows, test information leaks into training and the score looks better than it really is.
2. A new employee scored tomorrow is, by definition, not in the data the pipeline learned from, so the honest test is data the pipeline never saw.
3. Putting everything inside `Pipeline.fit()` (and inside every cross-validation fold) guarantees this; I proved it: the learned pay medians come from the 1,906 training rows and the 477 test rows influenced none of them.

### Stage 11 — Think for yourself: one new feature an HR manager would find sensible, and why it might help

**Feature: `pay_vs_dept_median`** = my pay divided by the median pay of colleagues in the same department and job level (medians learned on training rows only).

**Why it might help:** people compare themselves with peers, not with a company-wide number. Someone paid below their peers has a reason to look elsewhere; a raw salary of 8 lakh means very different things in different roles.

**Evidence on training rows:** the top quartile (paid above 1.10x peers) has the lowest leaver rate (10.3%) while the three lower quartiles leave at 15.3%, 17.2% and 15.7%. A weak but sensible pattern.

**Other engineered features:** `raise_recency_bucket` (0-6, 6-12, 12-18, 18+ months, or no raise on record): people whose last raise is 18+ months old leave at 31.7% versus 9-13% for everyone else, the strongest engineered signal. `never_promoted` (explains the blank `months_since_last_promotion`, 62% of rows). `overtime_flag`: honest note, 81% of rows have zero overtime, so its training 75th percentile is 0 and the flag really means "does any overtime" (18% of rows).

### Stages 9–10 — Evaluation, tuning

**What the numbers say.** With attendance and payroll aggregates the logistic regression scores ROC-AUC 0.710; with the attrition-table features only it scores 0.716, so the monthly aggregates add nothing here because only 383 of 2,383 rows (16%, 50 leavers) have them. The window sweep (1, 3 or 6 months: 0.645, 0.649, 0.680) is based on those 383 rows only and is noise-level, so I keep the brief's 3-month window.

### Stage 12 — Conclusions

**Conclusions.** `build_pipeline()` turns 43 raw columns into 142 model-ready columns, is fitted on training rows only, handles unseen categories and missing input columns, and is reused unchanged for scoring. The join keeps 2,383 rows (one per employee snapshot), drops nothing and uses only complete months that ended on or before the snapshot. **Recommendations:** get monthly history for the 2,000 historical rows; keep the date cutoff; reuse the same pipeline code for scoring so training and scoring never diverge.
